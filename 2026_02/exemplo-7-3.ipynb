```python
import pandas as pd

resultados_df = pd.read_csv("results.csv")
pilotos_df = pd.read_csv("drivers.csv")
```

```python
# filtrar colunas relevantes
colunas_pilotos = ["driverId", "forename", "surname", "nationality"]
pilotos_df = pilotos_df[colunas_pilotos]
colunas_pilotos_renomeados = {
    "driverId": "id_piloto",
    "forename": "nome",
    "surname": "sobrenome",
    "nationality": "nacionalidade"
}
pilotos_df = pilotos_df.rename(columns=colunas_pilotos_renomeados)
pilotos_df["nome_completo"] = pilotos_df["nome"] + " " + pilotos_df["sobrenome"]
pilotos_df.head()
```

```python
# resultados relevantes
resultados_df.head()
colunas_resultados = ["raceId", "driverId", "positionOrder"]
resultados_df = resultados_df[colunas_resultados]
colunas_resultados_renomeados = {
    "raceId": "id_corrida",
    "driverId": "id_piloto",
    "positionOrder": "posicao_final"
}
resultados_df = resultados_df.rename(columns=colunas_resultados_renomeados)
resultados_df.head()
```

```python
# filtrar somente vencedores (posição final = 1)
eh_vencedor = resultados_df["posicao_final"] == 1
vencedores_df = resultados_df[eh_vencedor]
vencedores_df.head()
```

```python
# mesclar os DataFrames de resultados e pilotos
vencedores_com_pilotos_df = pd.merge(
    vencedores_df, 
    pilotos_df,
    on="id_piloto",
)
vencedores_com_pilotos_df.head()
vencedores_com_pilotos_df.shape
```

```python
# Agrupar por quantidade de vitórias
agrupado_por_vitorias_df = vencedores_com_pilotos_df.groupby(
    ["id_piloto", "nome_completo", "nacionalidade"]
).size().reset_index(name="vitorias")
# ordenar por mais vitorias - descendente
agrupado_por_vitorias_df = agrupado_por_vitorias_df.sort_values(
    by="vitorias", ascending=False
)
agrupado_por_vitorias_df.head(10)
```
