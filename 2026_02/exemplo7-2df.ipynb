Criar funções para manipular arquivos com pandas

````python
import pandas as pd

def ler_arquivo(caminho: str) -> pd.DataFrame:
    try:
        conteudo_df = pd.read_csv(caminho)
        return conteudo_df

    except:
        print(f"Arquivo vazio ou corrompido. Criando um nova Base de dados.")
        colunas = ["codigo", "nome", "quantidade", "preco"]
        return pd.DataFrame(columns=colunas)

def salvar_arquivo(conteudo_df: pd.DataFrame, caminho: str) -> None:
    conteudo_df.to_csv(caminho, index=False)
````

Criar funções para manipular dados:
CRUD:
* Create - criar, inserir novo registro
* Read - Ler um registro existente
* Update - Atualizar o valor de um registro
* Delete - Remover um registro

````python
from IPython.display import display

def mostrar_estoque(estoque: pd.DataFrame, linhas=100) -> None:
    print("\n==== Mostrando Estoque ====\n")

    if estoque.empty:
        print("⚠️ O estoque está vazio no momento.")
    else:
        display(estoque.head(linhas))

    print("\n==== Fim do Estoque ====\n")


def cadastrar_peca(estoque_df: pd.DataFrame) -> pd.DataFrame:
    print("\n==== Cadastrando Nova Peça ====\n")
    codigo = input("Informe o código: ")
    nome = input("Informe o nome: ")
    quantidade = int(input("Informe a quantidade em estoque: "))
    preco = float(input("Informe o preço unitário: "))

    print("\n==== Adicionado ao banco de dados ====\n")
    nova_peca = {
        "codigo": codigo,
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }
    temp_nova_peca_df = pd.DataFrame([nova_peca])
    # concatenar o novo registro ao DataFrame existente
    # Concat junta os DataFrames, e ignore_index=True garante que o índice seja reindexado
    estoque_df = pd.concat(
        [estoque_df, temp_nova_peca_df], 
        ignore_index=True,
    )
    return estoque_df

def vender_peca(estoque_df: pd.DataFrame) -> pd.DataFrame:
    print("\n==== Vendendo peça do estoque ====\n")
    codigo = input("Informe o código da peça vendida: ")
    quantidade_vendida = int(input("Informe a quantidade de peças vendidas: "))

    filtro_codigo = estoque_df["codigo"] == str(codigo)

    if estoque_df[filtro_codigo].empty:
        print(f"\nErro: Peça com o código '{codigo}' não foi encontrada no estoque!")
        return estoque_df
    
    quantidade_estoque = estoque_df.loc[filtro_codigo, "quantidade"].values[0]

    estoque_pos_venda = quantidade_estoque - quantidade_vendida

    if estoque_pos_venda < 0:
        print("\nPeça não pode ser vendida - sem estoque suficiente!!!")
        print(f"Quantidade desejada: {quantidade_vendida}")
        print(f"Quantidade disponível: {quantidade_estoque}")
    else:
        estoque_df.loc[filtro_codigo, "quantidade"] = estoque_pos_venda
        print("\nVenda realizada com sucesso!!!")
        print(f"Novo estoque da peça '{codigo}': {estoque_pos_venda} unidades.")

    return estoque_df

def remover_peca(estoque_df: pd.DataFrame) -> pd.DataFrame:
    print("\n==== Removendo peça do estoque ====\n")
    codigo = input("Informe o código da peça a ser removida: ")

    filtro_codigo = estoque_df["codigo"] == codigo

    if estoque_df[filtro_codigo].empty:
        print(f"\nErro: Peça com o código '{codigo}' não foi encontrada no estoque!")
        return estoque_df

    # Remover a peça do DataFrame
    estoque_df = estoque_df[~filtro_codigo]
    print(f"\nPeça com o código '{codigo}' removida do estoque com sucesso!")

    return estoque_df
````

Criar menu de opções e processamento

````python
def menu_opcoes() -> None:
    print("=" * 40)
    print("Digite a opção desejada, conforme abaixo:")
    print("1 - Consultar Estoque")
    print("2 - Cadastrar nova peça")
    print("3 - Vender peça do estoque")
    print("4 - Remover peça do estoque")
    print("=" * 40)
    print("Para sair do programa, digite qualquer outra tecla.")
    print("=" * 40)
    # opcao_usuario = input("Entre com a opção desejada: ")
    # return opcao_usuario

def processar_opcao(opcao_usuario: str, estoque: pd.DataFrame) -> tuple[pd.DataFrame, bool]:
    continuar = True
    if opcao_usuario == "1":
        mostrar_estoque(estoque)
    elif opcao_usuario == "2":
        estoque = cadastrar_peca(estoque)
    elif opcao_usuario == "3":
        estoque = vender_peca(estoque)
    elif opcao_usuario == "4":
        estoque = remover_peca(estoque)
    else:
        print("Opção inválida. Encerrando o programa.")
        continuar = False
    return estoque, continuar
````

Criar execução principal do programa

````python
from IPython.display import clear_output

caminho_arquivo = "estoque.csv"
estoque_df = ler_arquivo(caminho_arquivo)
clear_output(wait=True)
print()

print("=" * 40)
print("Programa de controle de estoque")
print("=" * 40)
input("Pressione Enter para continuar...")

while True:
    clear_output(wait=True)
    print()

    menu_opcoes()
    opcao_usuario = input("Entre com a opção desejada: ")
    estoque_df, continuar = processar_opcao(opcao_usuario, estoque_df)

    input("\nPressione Enter para continuar...")
    salvar_arquivo(estoque_df, caminho_arquivo)

    if not continuar:
        break
````
