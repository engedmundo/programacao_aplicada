Criar funções para manipular arquivos com pandas

````python
import json


def ler_arquivo(caminho: str) -> dict:
    try:
        with open(caminho, mode="r+", encoding="utf-8") as arquivo:
            conteudo = json.load(arquivo)

        return conteudo

    except:
        print(f"Erro ao ler o arquivo {caminho}. Criando nova base de dados...")
        return {}

def salvar_arquivo(conteudo: dict, caminho: str) -> None:
    with open(caminho, mode="w", encoding="utf-8") as arquivo:
        json.dump(conteudo, arquivo, indent=4, ensure_ascii=False)

````

Exemplo de arquivo json salvo:
```json
estoque = [
    {
        "codigo": "Código da Peça", 
        "nome": "Nome da Peça", 
        "quantidade": "Quantidade da Peça", 
        "preco": "Valor da Peça", 
    }
]
```

Criar funções para manipular dados:
CRUD:
* Create - criar, inserir novo registro
* Read - Ler um registro existente
* Update - Atualizar o valor de um registro
* Delete - Remover um registro

````python
def mostrar_estoque(estoque: dict) -> None:
    separador = "=" * 63
    cabecalho = "| Código | Nome       | Qtd | Preço (R$) |"
    print(separador)
    print(cabecalho)

    for codigo, info in estoque.items():
        print(f"| {codigo:<6}", end=" | ")
        print(f"{info['nome']:<10}", end=" | ")
        print(f"{info['quantidade']:>3}", end=" | ")
        print(f"{info['preco']:>5}", end=" |\n")
    
    print(separador)


def cadastrar_peca(estoque: dict) -> dict:
    print("\n==== Cadastrando Nova Peça ====\n")
    codigo = input("Informe o código: ")
    nome = input("Informe o nome: ")
    quantidade = int(input("Informe a quantidade em estoque: "))
    preco = float(input("Informe o preço unitário]: "))

    print("\n==== Adicionado ao banco de dados ====\n")
    nova_peca = {
        "nome": nome,
        "quantidade": quantidade,
        "preco": preco
    }
    estoque[codigo] = nova_peca

    return estoque

def vender_peca(estoque: dict) -> dict:
    print("\n==== Vendendo peça do estoque ====\n")
    codigo = input("Informe o código da peça vendida: ")
    quantidade_vendida = int(input("Informe a quantidade de peças vendidas: "))

    peca = estoque[codigo]
    quantidade_estoque = peca["quantidade"]

    estoque_pos_venda = quantidade_estoque - quantidade_vendida

    if estoque_pos_venda < 0:
        print("Peça não pode ser vendida - sem estoque suficiente!!!")
        print(f"Quantidade desejada: {quantidade_vendida}")
        print(f"Quantidade disponível: {quantidade_estoque}")
    
    else:
        print("Venda realizada com sucesso!!!")
        # atualizar a quantidade de peças em estoque
        peca["quantidade"] = estoque_pos_venda
        # atualizar o registro da peça no estoque
        estoque[codigo] = peca

    return estoque

def remover_peca(estoque: dict) -> dict:
    print("\n==== Removendo peça do estoque ====\n")
    codigo = input("Informe o código da peça a ser removida: ")

    if codigo in estoque:
        del estoque[codigo]
        print(f"Peça com código '{codigo}' removida do estoque.")
    else:
        print(f"Erro: Peça com código '{codigo}' não encontrada no estoque!")

    return estoque
````

Criar menu de opções e processamento

````python
def menu_opcoes() -> None:
    print("=" * 40)
    print("Digite a opção desejada, conforme abaixo:")
    print("1 - Consultar Estoque")
    print("2 - Cadastrar nova peça")
    print("3 - Vender peça do estoque")
    print("4 - Remover peça do estoque")
    print("=" * 40)
    print("Para sair do programa, digite qualquer outra tecla.")
    print("=" * 40)

def processar_opcao(opcao_usuario: str, estoque: dict) -> tuple[dict, bool]:
    continuar = True
    if opcao_usuario == "1":
        mostrar_estoque(estoque)
    elif opcao_usuario == "2":
        estoque = cadastrar_peca(estoque)
    elif opcao_usuario == "3":
        estoque = vender_peca(estoque)
    elif opcao_usuario == "4":
        estoque = remover_peca(estoque)
    else:
        print("Opção inválida. Encerrando o programa.")
        continuar = False
    return estoque, continuar
````

Criar execução principal do programa

````python
from IPython.display import clear_output

caminho_arquivo = "estoque.json"
estoque = ler_arquivo(caminho_arquivo)
clear_output(wait=True)
print()

print("=" * 40)
print("Programa de controle de estoque")
print("=" * 40)
input("Pressione Enter para continuar...")

while True:
    clear_output(wait=True)
    print()

    menu_opcoes()
    opcao_usuario = input("Entre com a opção desejada: ")
    estoque, continuar = processar_opcao(opcao_usuario, estoque)

    input("\nPressione Enter para continuar...")
    salvar_arquivo(estoque, caminho_arquivo)

    if not continuar:
        break
````
